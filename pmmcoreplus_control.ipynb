```python

class mikroscope:
    def __init__(self):

        mikroscope.camera = mmc.getCameraDevice()
        mikroscope.DIAshutter = mmc.getShutterDevice()
        mikroscope.focus= mmc.getFocusDevice()
        mikroscope.stage = mmc.getXYStageDevice()
        mikroscope.PFS = mmc.getAutoFocusDevice()
        mikroscope.EPIshutter = 'TIEpiShutter'
        mikroscope.DIAlamp = 'TIDiaLamp'
        mikroscope.scope = 'TIScope'
        mikroscope.zoom = 'TINosePiece'
        mikroscope.filter = 'TIFilterBlock1'
        mikroscope.lightpath = 'TILightPath'
        mikroscope.PFS_offset = 'TIPFSOffset'
        mikroscope.core = 'Core'


        mikroscope.CamProperties = mmc.getDevicePropertyNames(mikroscope.camera)



        mikroscope.ScopeProperties = mmc.getDevicePropertyNames(mikroscope.scope)

    

        mikroscope.DIALampProperties = mmc.getDevicePropertyNames(mikroscope.DIAlamp)

       

        mikroscope.ZoomProperties = mmc.getDevicePropertyNames(mikroscope.zoom)
        print(mikroscope.ZoomProperties)
        

        mikroscope.zoom4x = '1-(Achromat) 4x NA 0.10 Dry'
        mikroscope.zoom10x = '2-(Achromat) 10x NA 0.25 Dry'
        mikroscope.zoom20x = '3-(Achromat) 20x NA 0.40 Dry'
        mikroscope.zoom40x = '4-S Plan Fluor 40x NA 0.60 Dry'
        mikroscope.zoom60x = '5-Plan Apo 60x NA 1.40 Oil'
        mikroscope.zoomempty = '6-Unknown'

        mikroscope.FilterProperties = mmc.getDevicePropertyNames(mikroscope.filter)

        

        mikroscope.filter_Cy5_1 = '1-Cy5'
        mikroscope.filter_Cy3_2 = '2-Cy3'
        mikroscope.filter_3 = '3----'
        mikroscope.filter_FITC_4 = '4-FITC'
        mikroscope.filter_5 = '5------'
        mikroscope.filter_DIA_6 = '6-DIA'


        mikroscope.CoreProperties = mmc.getDevicePropertyNames(mikroscope.core)


        mikroscope.DIAShutterProperties = mmc.getDevicePropertyNames(mikroscope.DIAshutter)


        mikroscope.EPIShutterProperties = mmc.getDevicePropertyNames(mikroscope.EPIshutter)


        mikroscope.StageProperties = mmc.getDevicePropertyNames(mikroscope.stage)


        mikroscope.LightPathProperties = mmc.getDevicePropertyNames(mikroscope.lightpath)
        
        mikroscope.eyepath = '1-Eye100'
        mikroscope.camerapath = '3-Right100'

        # Camera Variables
        mikroscope.temperature70 = 1
        mikroscope.exposuretext = '50'
        mikroscope.eyepathlight = 1
        mikroscope.camerapathlight = 1
        mikroscope.gaintext = '10'
        mikroscope.EMslider = 5.1
        mikroscope.EMtext = '5.1x'

        # Zoom Variables
        mikroscope.zoomlight4x = 0
        mikroscope.zoomlight10x = 0
        mikroscope.zoomlight20x = 0
        mikroscope.zoomlight40x = 0
        mikroscope.zoomlight60x = 0

        # Initialisation
        mmc.setProperty(mikroscope.camera,'CCDTemperatureSetPoint',-70)
        #set(mikroscope.temperature70,'Value', 1)
        mmc.setProperty(mikroscope.camera,'Exposure', 50)
        #set(mikroscope.exposuretext,'String', '50')
        mmc.setProperty(mikroscope.camera,'Binning', 2)
        mmc.setProperty(mikroscope.lightpath,'Label', mikroscope.eyepath)
        set(mikroscope.eyepathlight,'Value', 1)
        set(mikroscope.camerapathlight,'Value', 1)
        mmc.setProperty(mikroscope.camera,'Gain', 10)
        set(mikroscope.gaintext,'String', '10')
        mmc.setProperty(mikroscope.camera,'Pre-Amp-Gain', '5.1x')
        set(mikroscope.EMslider, 'Value', 5.1)
        set(mikroscope.EMtext,'String', '5.1x')
        mmc.setProperty(mikroscope.camera,'PixelType', '16bit')
        # mmc.setPosition(4500)
        # mmc.setProperty(mikroscope.DIAlamp,'State',1)


        # Zoom INI
        zoom_ini = mmc.getProperty(mikroscope.zoom,'Label')
        print(zoom_ini)
        if zoom_ini == mikroscope.zoom4x:
            set(mikroscope.zoomlight4x,'Value',1)
        elif zoom_ini == mikroscope.zoom10x:
            set(mikroscope.zoomlight10x,'Value',1)
        elif zoom_ini == mikroscope.zoom20x:
            set(mikroscope.zoomlight20x,'Value',1)    
        elif zoom_ini == mikroscope.zoom40x:
            set(mikroscope.zoomlight40x,'Value',1)
        elif zoom_ini == mikroscope.zoom60x:
            set(mikroscope.zoomlight60x,'Value',1)

        # Light Path INI
        mmc.setProperty(mikroscope.lightpath,'Label',mikroscope.eyepath)
        set(mikroscope.eyepathlight,'Value',1)
        set(mikroscope.camerapathlight,'Value',0)
mikroscope()
print(mikroscope.stage)
```
